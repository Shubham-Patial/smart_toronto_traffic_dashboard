````python
def map_weather(row):
    if 'Loose Snow' in str(row['RDSFCOND']) or 'Slush' in str(row['RDSFCOND']) or 'Packed Snow' in str(row['RDSFCOND']):
        return 'Snowy'
    elif 'Wet' in str(row['RDSFCOND']) or 'Spilled liquid' in str(row['RDSFCOND']):
        return 'Rainy'
    elif 'Clear' in str(row['VISIBILITY']):
        return 'Clear'
    elif 'Fog' in str(row['VISIBILITY']) or 'Mist' in str(row['VISIBILITY']):
        return 'Foggy'
    else:
        return 'Other'

df['weather_condition'] = df.apply(map_weather, axis=1)
````

````python
df['weather_condition']
````

````python
print(df["VEHTYPE"].unique())

def simplify_vehicle(v):
    v = str(v)
    if 'Automobile' in v or 'Car' in v:
        return 'Car'
    elif 'Police Vehicle' in v:
        return 'Police Car'
    elif 'Truck' in v:
        return 'Truck'
    elif 'Motorcycle' in v:
        return 'Motorcycle'
    elif 'Bicycle' in v or 'Cyclist' in v:
        return 'Bicycle'
    elif 'Bus' in v:
        return 'Bus'
    else:
        return 'Other'

df['vehicle_category'] = df['VEHTYPE'].apply(simplify_vehicle)
````

````python
print(df["INJURY"].unique())
injury_map = {
    'Fatal': 3,
    'Major': 2,
    'Major Injury': 2,
    'Minor': 1,
    'Non-Fatal Injury': 1,
    'None': 0
}

df['injury_severity'] = df['INJURY'].map(injury_map).fillna(0)
````
